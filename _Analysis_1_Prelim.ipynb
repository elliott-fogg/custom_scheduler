## Do DIRECT Requests have Locations?
No.

```python
direct_group_ids = df_group[df_group["observation_type"]=="DIRECT"]["id"].unique().tolist()
direct_request_ids = df_request[df_request["request_group_id"].isin(direct_group_ids)]["id"].unique().tolist()
direct_locations = df_location[df_location["request_id"].isin(direct_request_ids)]
direct_locations
```

## Observation Type Split
Looking at number of both RequestGroups and Requests.

```python
obstype_group_count = df_group["observation_type"].value_counts()
obstype_group_perc = obstype_group_count/len(df_group)*100
temp = df_request.merge(df_group[["id", "observation_type"]], left_on="request_group_id", right_on="id")
obstype_req_count = temp["observation_type"].value_counts()
obstype_req_perc = obstype_req_count/len(df_request)*100
pd.concat([obstype_group_count, obstype_group_perc, obstype_req_count, obstype_req_perc], axis=1)\
    .set_axis(["RequestGroup Count", "RequestGroup %", "Request Count", "Request %"], axis=1)
```
