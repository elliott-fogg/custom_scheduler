```python
import pandas as pd
import pickle
import os
```

```python
def calculate_total_priority(output_data, perfect=False):
    if perfect:
        data = output_data["scheduled"]
    else:
        data = output_data["final_completed_requests"]

    total_priority = 0
    for request_id, values in data.items():
        total_priority += values["priority"]
    return total_priority
```

```python
for item in os.walk("output_files"):
    # print(item)
    for filename in item[2]:
        print(filename)
```

```python
d1 = pickle.load(open("output_files/perfect/perfect_2020-08-01.pkl", "rb"))
d2 = pickle.load(open("output_files/perfect/perfect_2020-08-01._perfect.pkl", "rb"))
print("Original:", calculate_total_priority(d1))
print("Perfect:", calculate_total_priority(d2, perfect=True))

s1 = set(d1["final_completed_requests"].keys())
s2 = set(d2["scheduled"].keys())
si = s1.intersection(s2)
s1_o = s1 - si
s2_o = s2 - si
print("Requests scheduled in original:", len(s1))
print("Requests scheduled in Perfect:", len(s2))
print("Requests Lost/Gained:", len(s1_o),"/",len(s2_o))
# print(len(s1), len(s2), len(si), len(s1_o), len(s2_o))
```
