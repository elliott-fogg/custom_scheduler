```python
input_uuid_filepath = "input_files/input_uuids.json"
if os.path.isfile(input_uuid_filepath):
    input_uuids = json.load(open(input_uuid_filepath, "r"))
else:
    input_uuids = {}

for filename, filepath in input_files.items():
    data = pickle.load(open(filepath, "rb"))
    if "dataset_id" not in data:
        while True:
            dataset_id = str(uuid4())
            if dataset_id not in input_uuids.values():
                break
        print("Setting new UUID:", filename, dataset_id)
        data["dataset_id"] = dataset_id
        pickle.dump(data, open(filepath, "wb"))
        input_uuids[filename] = dataset_id

    else:
        print("Using old UUID:", filename, data["dataset_id"])
        data["dataset_id"] = str(data["dataset_id"])
        pickle.dump(data, open(filepath, "wb"))
        input_uuids[filename]  = str(data["dataset_id"])

with open(input_uuid_filepath, "w") as uuidf:
    json.dump(input_uuids, uuidf, indent=4)
```

## Add corresponding UUIDs to Output Files

```python
input_uuids = json.load(open(input_uuid_filepath, "r"))

for item in os.walk("output_files"):
    path = item[0]
    for filename in item[2]:
        if filename in input_uuids:
            output_uuid = input_uuids[filename]
            filepath = os.path.join(path, filename)
            data = pickle.load(open(filepath, "rb"))
            if "dataset_id" in data:
                if data["dataset_id"] == output_uuid:
                    print(filename, "- Correct UUID already present.")
                else:
                    print(filename, "- UNMATCHING UUID FOUND!")
                    print(f"Input: {output_uuid}, Output: {data['dataset_id']}")
            else:
                data["dataset_id"] = output_uuid
                pickle.dump(data, open(filepath, "wb"))
                print("Added UUID to", filename)
            
        else:
            print("NOT FOUND - ", filename)
```
