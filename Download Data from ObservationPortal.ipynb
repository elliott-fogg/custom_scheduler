```python
import requests
import json
import os
import math
```

```python
output_folder = "D:/_Observation_Portal_Data"
os.path.isdir(output_folder)
```

```python
count = 250
requests_per_file = 1000
offset = count*requests_per_file

next_url = f"http://localhost:8000/api/requestgroups/?format=json&limit={requests_per_file}&offset={offset}"

while next_url != None:
    r = requests.get(next_url).json()
    results = r["results"]
    filepath = os.path.join(output_folder, f"observation_portal_data_{count}.json")
    with open(filepath, "w") as f:
        json.dump(results, f)
    count += 1

    next_url = r["next"]
        
    print(f"\r{count} / {math.floor(r['count']/requests_per_file)+1}", end="")
```
